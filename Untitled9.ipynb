### Distribution of Flight Prices

First, let's visualize the distribution of the `Price` column. This is what a 'normal curve' or distribution plot typically shows for a single variable. It helps us understand the spread and central tendency of the prices.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Price'], kde=True, bins=50)
plt.title('Distribution of Flight Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Checking for Linear or Polynomial Relationships

To check for linear or polynomial relationships between variables (e.g., between `Duration` and `Price`), a scatter plot with a regression line is much more appropriate. Before we can do that, the `Duration` column needs to be converted from a string format (e.g., '2h 50m') into a numerical format (e.g., total minutes).

```python
def convert_duration_to_minutes(duration_str):
    if pd.isna(duration_str): # Handle NaN values if any
        return np.nan
    parts = duration_str.split(' ')
    total_minutes = 0
    for part in parts:
        if 'h' in part:
            total_minutes += int(part.replace('h', '')) * 60
        elif 'm' in part:
            total_minutes += int(part.replace('m', ''))
    return total_minutes

df['Duration_minutes'] = df['Duration'].apply(convert_duration_to_minutes)

print(df[['Duration', 'Duration_minutes']].head())
```

Now that `Duration` is in a numerical format (`Duration_minutes`), we can create a scatter plot with a regression line to observe its relationship with `Price`. A straight line suggests a linear relationship, while a curve might indicate a polynomial relationship.

```python
plt.figure(figsize=(12, 7))
sns.regplot(x='Duration_minutes', y='Price', data=df, scatter_kws={'alpha':0.3})
plt.title('Relationship between Flight Duration and Price')
plt.xlabel('Duration (minutes)')
plt.ylabel('Price')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Outlier Removal

Outliers can significantly skew statistical analysis and model performance. We will use the Interquartile Range (IQR) method to identify and remove outliers from numerical columns. The IQR method defines outliers as data points that fall below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR`, where Q1 is the first quartile, Q3 is the third quartile, and IQR is the difference between Q3 and Q1.

```python
# Function to remove outliers using IQR
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df_filtered = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    return df_filtered

print(f"Shape of DataFrame before outlier removal: {df.shape}")

# Remove outliers from 'Price' column
df_filtered = remove_outliers_iqr(df, 'Price')
print(f"Shape of DataFrame after removing 'Price' outliers: {df_filtered.shape}")

# Remove outliers from 'Duration_minutes' column (if it exists and is numeric)
if 'Duration_minutes' in df_filtered.columns and pd.api.types.is_numeric_dtype(df_filtered['Duration_minutes']):
    df_filtered = remove_outliers_iqr(df_filtered, 'Duration_minutes')
    print(f"Shape of DataFrame after removing 'Duration_minutes' outliers: {df_filtered.shape}")

df = df_filtered.copy()
print(f"Final shape of DataFrame after all outlier removal: {df.shape}")
```

```python
plt.figure(figsize=(12, 7))
sns.regplot(x='Duration_minutes', y='Price', data=df, scatter_kws={'alpha':0.3})
plt.title('Relationship between Flight Duration and Price')
plt.xlabel('Duration (minutes)')
plt.ylabel('Price')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
